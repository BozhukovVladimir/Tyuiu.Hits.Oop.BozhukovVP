```
public interface IAwardWinning
{
    int AwardsCount { get; set; }
    void DisplayAwards();
    string GetAwardLevel(); 
}

public interface IAwardService
{
    void RegisterAward(string awardName, int year);
    int GetTotalAwardsCount();
    List<string> GetAwardHistory();
}

public class AwardService : IAwardService
{
    private List<(string Name, int Year)> _awards = new List<(string, int)>();
    
    public void RegisterAward(string awardName, int year)
    {
        _awards.Add((awardName, year));
        Console.WriteLine($"Зарегистрирована награда: {awardName} ({year} год)");
    }
    
    public int GetTotalAwardsCount() => _awards.Count;
    
    public List<string> GetAwardHistory()
    {
        return _awards.Select(a => $"{a.Name} ({a.Year})").ToList();
    }
}

public class Movie
{
    private string _productionStudio;
    private List<string> _filmingLocations;
    private bool _isRestored;

    public string ProductionStudio
    {
        get => _productionStudio;
        set => _productionStudio = !string.IsNullOrWhiteSpace(value) ? value : "Неизвестно";
    }

    public bool IsRestored 
    {
        get => _isRestored;
        set => _isRestored = value;
    }

    public IReadOnlyList<string> FilmingLocations => _filmingLocations.AsReadOnly();

    public Movie(string title, int year, string director, int duration)
    {
        _filmingLocations = new List<string>();
        ProductionStudio = "Неизвестно";
        IsRestored = false;
    }

    public void AddFilmingLocation(string location)
    {
        if (!string.IsNullOrWhiteSpace(location))
        {
            _filmingLocations.Add(location);
            Console.WriteLine($"Добавлена локация съёмок: {location}");
        }
    }

    public virtual void RestoreFilm()
    {
        if (!IsRestored)
        {
            IsRestored = true;
            Console.WriteLine($"Фильм «{Title}» был отреставрирован");
        }
    }

    public string GetFilmingLocationsInfo()
    {
        if (!_filmingLocations.Any())
            return "Локации съёмок не указаны";
        
        return $"Локации съёмок: {string.Join(", ", _filmingLocations)}";
    }

    public virtual string GetInfo()
    {
        string restored = IsRestored ? " [РЕСТАВРИРОВАН]" : "";
        return $"«{Title}» ({Year}), реж. {Director}, {Duration} мин, {Country}{restored}";
    }
}

public class FeatureFilm : Movie, IAwardWinning
{
    private string _screenplayAuthor;
    private bool _hasSequel;

    public string ScreenplayAuthor
    {
        get => _screenplayAuthor;
        set => _screenplayAuthor = !string.IsNullOrWhiteSpace(value) ? value : "Неизвестно";
    }

    public bool HasSequel
    {
        get => _hasSequel;
        set => _hasSequel = value;
    }

    private readonly IAwardService _awardService;

    public FeatureFilm(string title, int year, string director, string genre, 
        double budget, int duration, IAwardService awardService) 
        : base(title, year, director, duration)
    {
        Genre = genre;
        Budget = budget;
        AwardsCount = 0;
        ProductionCompany = "Неизвестно";
        _awardService = awardService;
        HasSequel = false;
        ScreenplayAuthor = director; 
    }

    // Новые методы
    public void AnnounceSequel(string sequelTitle)
    {
        if (!HasSequel)
        {
            HasSequel = true;
            Console.WriteLine($"Анонсирован сиквел «{sequelTitle}» к фильму «{tle}»!");
        }
    }

    public void RegisterAward(string awardName, int year)
    {
        _awardService.RegisterAward(awardName, year);
        AwardsCount = _awardService.GetTotalAwardsCount();
    }

    public void ShowAwardHistory()
    {
        var history = _awardService.GetAwardHistory();
        if (history.Any())
        {
            Console.WriteLine($"История наград фильма «{Title}»:");
            foreach (var award in history)
            {
                Console.WriteLine($"  - {award}");
            }
        }
    }

    // Реализация нового метода интерфейса
    public string GetAwardLevel()
    {
        return AwardsCount switch
        {
            0 => "Без наград",
            < 3 => "Мало наград",
            < 10 => "Умеренное количество наград",
            _ => "Много наград"
        };
    }

    public override string GetInfo()
    {
        string sequelInfo = HasSequel ? " [ЕСТЬ СИКВЕЛ]" : "";
        return base.GetInfo() + $", жанр: {Genre}, бюджет: ${Budget} млн{sequelInfo}";
    }
}

// Дополненный Documentary с явной реализацией интерфейса
public class Documentary : Movie, IAwardWinning
{
    // Существующие поля...
    private int _interviewCount;
    private string _scientificAdvisor;

    public int InterviewCount
    {
        get => _interviewCount;
        set => _interviewCount = value >= 0 ? value : 0;
    }

    public string ScientificAdvisor
    {
        get => _scientificAdvisor;
        set => _scientificAdvisor = !string.IsNullOrWhiteSpace(value) ? value : "Не указан";
    }

    // Явная реализация интерфейса
    private int _awardsCount;
    int IAwardWinning.AwardsCount 
    { 
        get => _awardsCount;
        set => _awardsCount = value >= 0 ? value : 0;
    }

    void IAwardWinning.DisplayAwards()
    {
        string type = IsEducational ? "Научный" : "Документальный";
        Console.WriteLine($"{type} фильм «{Title}» получил {_awardsCount} наград (явная реализация)");
    }

    string IAwardWinning.GetAwardLevel()
    {
        return _awardsCount switch
        {
            0 => "Без научных наград",
            < 5 => "Признанный научный фильм",
            _ => "Высоко оценённый научный труд"
        };
    }

    public Documentary(string title, int year, string director, string topic, 
        bool isEducational, int duration) 
        : base(title, year, director, duration)
    {
        Topic = topic;
        IsEducational = isEducational;
        InterviewCount = 0;
        ScientificAdvisor = "Не указан";
    }

    // Новые методы
    public void AddInterview()
    {
        InterviewCount++;
        Console.WriteLine($"Добавлено интервью #{InterviewCount} для фильма «{Title}»");
    }

    public void SetScientificAdvisor(string advisor)
    {
        ScientificAdvisor = advisor;
        Console.WriteLine($"Научный консультант установлен: {advisor}");
    }

    public override void RestoreFilm()
    {
        if (!IsRestored && Year < 2000)
        {
            base.RestoreFilm();
            Console.WriteLine("Особое внимание уделено архивным материалам");
        }
    }

    public override string GetInfo()
    {
        string edu = IsEducational ? "образовательный" : "популярный";
        string interviews = InterviewCount > 0 ? $", интервью: {InterviewCount}" : "";
        return base.GetInfo() + $", тема: {Topic} ({edu}{interviews})";
    }
}

// Сервис для управления коллекциями
public interface IFilmCollectionService
{
    void AddToFestival(Movie movie);
    void AddToArchive(Movie movie);
    List<Movie> GetRecommendedFilms();
}

public class FilmCollectionService : IFilmCollectionService
{
    private List<Movie> _festivalFilms = new List<Movie>();
    private List<Movie> _archiveFilms = new List<Movie>();

    public void AddToFestival(Movie movie)
    {
        _festivalFilms.Add(movie);
        Console.WriteLine($"Фильм «{movie.Title}» добавлен в фестивальную программу");
    }

    public void AddToArchive(Movie movie)
    {
        if (movie.Year < DateTime.Now.Year - 10)
        {
            _archiveFilms.Add(movie);
            Console.WriteLine($"Фильм «{movie.Title}» перемещен в архив");
        }
    }

    public List<Movie> GetRecommendedFilms()
    {
        return _festivalFilms
            .Where(f => f.CalculateRating() >= 7.0)
            .ToList();
    }
}

// Обновленный код демонстрации
var awardService = new AwardService();
var collectionService = new FilmCollectionService();

var interstellar = new FeatureFilm("Интерстеллар", 2014, "Кристофер Нолан", 
    "фантастика", 165, 169, awardService);

var earthlings = new Documentary("Земляне", 2005, "Шон Монсон", 
    "права животных", true, 95);

// Настройка фильмов
interstellar.AddFilmingLocation("Лос-Анджелес");
interstellar.AddFilmingLocation("Исландия");
interstellar.ScreenplayAuthor = "Джонатан Нолан";
interstellar.ProductionStudio = "Warner Bros";

earthlings.AddFilmingLocation("США");
earthlings.AddInterview();
earthlings.AddInterview();
earthlings.SetScientificAdvisor("Доктор Джейн Смит");

// Работа с наградами
interstellar.RegisterAward("Оскар", 2015);
interstellar.RegisterAward("BAFTA", 2015);
interstellar.RegisterAward("Сатурн", 2015);

// Явная реализация интерфейса
((IAwardWinning)earthlings).AwardsCount = 2;

// Управление коллекциями
collectionService.AddToFestival(interstellar);
collectionService.AddToArchive(earthlings);

Console.WriteLine("=== ДОПОЛНЕННАЯ ИНФОРМАЦИЯ ===");
Console.WriteLine(interstellar.GetInfo());
Console.WriteLine(earthlings.GetInfo());
Console.WriteLine(interstellar.GetFilmingLocationsInfo());
Console.WriteLine(earthlings.GetFilmingLocationsInfo());

Console.WriteLine("\n=== НАГРАДЫ И СЕРВИСЫ ===");
interstellar.ShowAwardHistory();
Console.WriteLine($"Уровень наград: {interstellar.GetAwardLevel()}");

// Явный вызов методов интерфейса
IAwardWinning awardEarthlings = earthlings;
awardEarthlings.DisplayAwards();
Console.WriteLine($"Уровень наград: {awardEarthlings.GetAwardLevel()}");

Console.WriteLine("\n=== РЕКОМЕНДАЦИИ ===");
var recommended = collectionService.GetRecommendedFilms();
foreach (var film in recommended)
{
    Console.WriteLine($"Рекомендуем: {film.GetInfo()}");
}

// Демонстрация новых возможностей
Console.WriteLine("\n=== НОВЫЕ ВОЗМОЖНОСТИ ===");
interstellar.AnnounceSequel("Интерстеллар 2");
earthlings.RestoreFilm();

var oldFilm = new Documentary("Древний мир", 1995, "Иван Петров", 
    "археология", true, 120);
oldFilm.RestoreFilm();
```
